```python
# Data Cleaning and preprocessing
```

```python
df.head()
```

```python
df = pd.get_dummies(df , columns=['ChestPainType'] , drop_first = True)
```

```python
# We cannot convert the entire DataFrame to int directly because columns like 'Sex', 'RestingECG', 'ExerciseAngina', and 'ST_Slope' contain text strings.
# Let's convert only the boolean and existing numeric columns to int, or convert other categorical columns first.

# Identify columns that can be converted to numeric/int safely
numeric_and_bool_cols = df.select_dtypes(include=['number', 'bool']).columns
df[numeric_and_bool_cols] = df[numeric_and_bool_cols].astype(int)
df.head()
```

```python
df_encode = pd.get_dummies(df,drop_first=True)
```

```python
df_encode
```

```python
df_encode = df_encode.astype(int)
```

```python
from sklearn.preprocessing import StandardScaler
columns = ['RestingBP' , 'Age' , 'Cholesterol' , 'MaxHR']
scaler = StandardScaler()
df_encode[columns] = scaler.fit_transform(df_encode[columns])
```
